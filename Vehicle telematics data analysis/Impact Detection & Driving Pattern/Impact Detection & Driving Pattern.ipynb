# Heatmap

```python
#Average Acceleration values by the hour of day and day of the week, 
#which can help us identify any patterns or trends in the data

import pandas as pd
import plotly.graph_objs as go

# Converting 'Time' and 'Date' columns to datetime format
df['time'] = pd.to_datetime(df['time'], format='%H:%M:%S').dt.time
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')

# Extracting hour and day of the week
df["hour"] = pd.to_datetime(df["time"], format='%H:%M:%S').dt.hour
df["day_of_week"] = df["date"].dt.day_name()

# Creating a pivot table for aggregation
agg_data = df.pivot_table(index="hour", columns="day_of_week", 
                          values=["x", "y", "z"], 
                          aggfunc="mean")

# Create a heatmap
fig = go.Figure(go.Heatmap(x=agg_data.columns.levels[1], 
                           y=agg_data.index, 
                           z=agg_data.values,
                           xgap=1, ygap=1, 
                           colorscale="RdBu", 
                           colorbar=dict(title="Average Acceleration")))
fig.update_layout(title="Average Acceleration by Hour of Day and Day of Week")
fig.show()
```

The heatmap visualization of average acceleration by the hour of the day and day of the week provides several insights:

1). Hourly Patterns: We can observe how average acceleration values change throughout the day. The x-axis represents the day of the week, and the y-axis represents the hour of the day. By looking at the color intensity in each cell, we can identify peak hours or time periods with higher or lower acceleration.

2). Day-of-Week Patterns: The heatmap allows us to compare acceleration patterns across different days of the week. It helps identify if there are consistent trends in acceleration for specific weekdays. **As from heatmap, we can notice higher acceleration during weekdays compared to weekends as maximum on "Thursday during 9:00 PM to 12:00 AM" and least on "Sunday from 12:00 AM to 5:00 AM".**

3). Peak Acceleration Hours: Darker cells in the heatmap indicate higher average acceleration during specific hours of specific days. These are potential peak hours of activity or acceleration events.

# Hard Brake Detection

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming your dataset has accelerometer data in columns 'x', 'y', and 'z'
# You can adjust these column names based on your dataset
accel_columns = ['x', 'y', 'z']

# Calculate the total acceleration as the magnitude of the accelerometer readings
df2['total_acceleration'] = (df2[accel_columns] ** 2).sum(axis=1) ** 0.5

# Set a threshold for hard braking detection (you can adjust this threshold as needed)
hard_brake_threshold = 0.987  # Adjust based on your data

# Detect hard braking events by checking if total acceleration exceeds the threshold
df2['hard_brake_event'] = df2['total_acceleration'] > hard_brake_threshold

# Visualize the accelerometer data and detected hard braking events
plt.figure(figsize=(12, 6))
plt.plot(df2.index, df2['total_acceleration'], label='Total Acceleration', color='blue')
plt.scatter(df2.index[df2['hard_brake_event']], df2['total_acceleration'][df2['hard_brake_event']], c='red', marker='o', label='Hard Brake Event')
plt.xlabel('Time')
plt.ylabel('Total Acceleration')
plt.title('Hard Braking Detection from Accelerometer Data')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

# Print the timestamps of hard braking events
hard_brake_events = df2[df2['hard_brake_event']]
print('Hard Braking Events:')
print(hard_brake_events[['timeStamp']])
```

1). Red Points on the Graph: The red points on the graph represent instances where "Hard Brake Events" have been detected. These events are marked with red dots on the graph.

2). Threshold for Hard Braking: A threshold, defined as hard_brake_threshold, is set to distinguish "Hard Brake Events" from normal driving. **If the "Total Acceleration" at a particular timestamp exceeds this threshold, it's considered a "Hard Brake Event."**

# Sharp Turns

```python
import pandas as pd

# Define thresholds for sharp turns based on angular acceleration
sharp_turn_threshold = 0.987  # Example threshold for sharp turns in degrees/s²

# Calculate the magnitude of angular acceleration
df2['angular_acceleration'] = (df2['x'] ** 2 + df2['y'] ** 2 + df2['z'] ** 2) ** 0.5

# Detect sharp turns
df2['event'] = 'Normal'
df2.loc[df2['angular_acceleration'] > sharp_turn_threshold, 'event'] = 'Sharp Turn'

# Print events
df2
```

**This is used to analyze accelerometer data and identify instances of "Sharp Turns" by comparing the angular acceleration to a specified threshold. It categorizes each data point as either a normal event or a sharp turn event, on the basis of that whether angular acceleration is greater than or less than the threshold value, allowing us to study and analyze driving patterns that involve significant changes in direction.**
